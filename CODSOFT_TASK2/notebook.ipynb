```python
import pandas as pd
import os

train_path = 'dataset/fraudTrain.csv'
test_path = 'dataset/fraudTest.csv'

if os.path.exists(train_path) and os.path.exists(test_path):
    print("Loading datasets. This might take a moment depending on file size...")
    train_df = pd.read_csv(train_path)
    test_df = pd.read_csv(test_path)
    
    print(f"Training data shape: {train_df.shape}")
    print(f"Testing data shape: {test_df.shape}")
    
    display(train_df.head())
    
    print("\nDataset Info (Training):")
    train_df.info()
    
    # The target column for this specific dataset is usually 'is_fraud'
    if 'is_fraud' in train_df.columns:
        print("\nClass Distribution (0 = Legitimate, 1 = Fraud):")
        print(train_df['is_fraud'].value_counts(normalize=True) * 100)
    elif 'Class' in train_df.columns:
        print("\nClass Distribution (0 = Legitimate, 1 = Fraud):")
        print(train_df['Class'].value_counts(normalize=True) * 100)
        
else:
    print("Files not found. Please ensure 'fraudTrain.csv' and 'fraudTest.csv' are exactly named and inside the 'dataset' folder.")
```

```python
print("Selecting important features...")
# Define the features we want to use and the target variable
features = ['category', 'amt', 'gender', 'city_pop', 'lat', 'long']
target = 'is_fraud'

# Extract these columns from both train and test datasets
X_train = train_df[features].copy()
y_train = train_df[target].copy()

X_test = test_df[features].copy()
y_test = test_df[target].copy()

print("Encoding categorical variables (category and gender)...")
# Convert categorical text data into numerical format using One-Hot Encoding
X_train = pd.get_dummies(X_train, columns=['category', 'gender'], drop_first=True)
X_test = pd.get_dummies(X_test, columns=['category', 'gender'], drop_first=True)

# Align the train and test sets to ensure they have the exact same columns after encoding
X_train, X_test = X_train.align(X_test, join='left', axis=1, fill_value=0)

print(f"Prepared Training Features shape: {X_train.shape}")
print(f"Prepared Testing Features shape: {X_test.shape}")
display(X_train.head())
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report

print("Scaling numerical features...")
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("Retraining Logistic Regression with balanced class weights...")
# Added class_weight='balanced' to handle the massive data imbalance
model = LogisticRegression(max_iter=1000, random_state=42, class_weight='balanced')

# Train the model
model.fit(X_train_scaled, y_train)

print("Making predictions on the test set...")
y_pred = model.predict(X_test_scaled)

accuracy = accuracy_score(y_test, y_pred)
print(f"\nTest Accuracy: {accuracy * 100:.2f}%\n")

print("Classification Report:")
print(classification_report(y_test, y_pred, target_names=['Legitimate (0)', 'Fraud (1)']))
```

```python
print("Saving predictions to CSV...")

# Add the predictions as a new column to the original test dataframe
test_df['Predicted_Fraud_Label'] = y_pred

# Select a few identifying columns along with our prediction to save
output_path = 'dataset/predicted_fraud.csv'
columns_to_save = ['trans_date_trans_time', 'cc_num', 'merchant', 'amt', 'Predicted_Fraud_Label']

test_df[columns_to_save].to_csv(output_path, index=False)

print(f"Task 2 Complete! Predictions successfully saved to {output_path}")
display(test_df[columns_to_save].head())
```
